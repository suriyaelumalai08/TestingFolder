```python
import matplotlib.pyplot as plt
```

```python
def mse_loss(pred,true):
    return ((pred - true)**2).mean()

y_true = np.array([1.0, 2.0, 3.0])
y_pred = np.array([1.2, 1.8, 3.5])
print("Loss:", mse_loss(y_pred, y_true).round(4)) 
```

```python
w = 5.0         
lr = 0.1          
history = []

for i in range(20):
    loss = w ** 2
    grad = 2 * w
    w = w- lr * grad
    history.append(round(loss,4))
print("Loss over steps:", history)
print("Final w:", round(w, 6)) 
```

```python
plt.plot(history, marker='o', markersize=4)
plt.xlabel('Step'); plt.ylabel('Loss')
plt.title('Gradient descent reducing loss'); plt.show()
```
