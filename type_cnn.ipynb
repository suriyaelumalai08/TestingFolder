```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import vgg16,vgg19
from tensorflow.keras.preprocessing import image
from tensorflow.keras import layers,models
import matplotlib.pyplot as plt
import tensorflow as tf
import numpy as np
import cv2
```

```python
IMG_SIZE = (224,224)
BATCH_SIZE = 32

train_dir = r'D:\python\python\testing\dataset\final\train'
test_dir = r'D:\python\python\testing\dataset\final\test'
```

```python
#DATA PERPROCESSING USE TO TRAIN AND TEST DATASET CREATION


train_gen = ImageDataGenerator(
    rescale = 1./255,
    shear_range = 0.2,
    zoom_range = 0.2,
    horizontal_flip = True
)

test_gen = ImageDataGenerator(rescale= 1./255)
```

```python
train_ds = train_gen.flow_from_directory(
    train_dir,
    batch_size = BATCH_SIZE,
    target_size = IMG_SIZE,
    class_mode = 'binary'
)

test_ds = test_gen.flow_from_directory(
    test_dir,
    batch_size = BATCH_SIZE,
    target_size = IMG_SIZE,
    class_mode = 'binary'
)
```

```python
base_model1 = vgg16.VGG16(weights='imagenet',include_top=True)

base_model2 = vgg19.VGG19(weights='imagenet',include_top=True)
```

```python
file_path = r'D:\python\python\testing\tffcar.jpg'
img = image.load_img(file_path,target_size=(224,224,3))
img_array = image.img_to_array(img)
img_array = np.expand_dims(img_array,axis=0)
img_array = vgg16.preprocess_input(img_array)
```

```python
predicition_vgg16 = base_model1.predict(img_array)
predicition_vgg19 = base_model2.predict(img_array)
```

```python
decoded_vgg16 = vgg16.decode_predictions(predicition_vgg16,top=10)
decoded_vgg19 = vgg19.decode_predictions(predicition_vgg19,top=10)
```

```python
print('Top prediction of the vgg16')
for i ,(imagenetId,label,prob) in enumerate(decoded_vgg16[0]):
    print(f'step:{i+1}, label: {label}, prob:{prob*100:.2f}%')
```

```python
print('Top prediction of the vgg19')
for i ,(imagenetId,label,prob) in enumerate(decoded_vgg19[0]):
    print(f'step:{i+1}, label: {label}, prob:{prob*100:.2f}%')
```

```python
import tensorflow as tf
tf.keras.backend.clear_session()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

# initstion the model
model = Sequential()
model.add(Conv2D(64, (3,3), padding='same', activation='relu', input_shape=(224,224,3)))
model.add(Conv2D(64, (3,3), padding='same', activation='relu'))
model.add(MaxPooling2D((2,2), strides=(2,2)))

model.add(Conv2D(128, (3,3), padding='same', activation='relu'))
model.add(Conv2D(128, (3,3), padding='same', activation='relu'))
model.add(MaxPooling2D((2,2), strides=(2,2)))

model.add(Conv2D(256, (3,3), padding='same', activation='relu'))
model.add(Conv2D(256, (3,3), padding='same', activation='relu'))
model.add(Conv2D(256, (3,3), padding='same', activation='relu'))
model.add(MaxPooling2D((2,2), strides=(2,2)))

model.add(Conv2D(512, (3,3), padding='same', activation='relu'))
model.add(Conv2D(512, (3,3), padding='same', activation='relu'))
model.add(Conv2D(512, (3,3), padding='same', activation='relu'))
model.add(MaxPooling2D((2,2), strides=(2,2)))

model.add(Conv2D(512, (3,3), padding='same', activation='relu'))
model.add(Conv2D(512, (3,3), padding='same', activation='relu'))
model.add(Conv2D(512, (3,3), padding='same', activation='relu'))
model.add(MaxPooling2D((2,2), strides=(2,2)))

model.add(Flatten())
model.add(Dense(128, activation='relu'))
model.add(Dense(1, activation='sigmoid'))  # Changed for binary classification

model.summary()

# result = model.predict(np.expand_dims(img,axis=0))
```

```python
myoptimizer= tf.keras.optimizers.Adam(learning_rate=0.01)
myloss = tf.keras.losses.BinaryCrossentropy()  # Changed for binary classification
model.compile(
    optimizer = myoptimizer,
    loss = myloss,
    metrics = ['accuracy']

)
```

```python
model.fit(train_ds,epochs=10)
```

```python
model.fit(train_ds, epochs=10, validation_data=test_ds)
```

```python
# Get class names from training data

class_names = list(train_ds.class_indices.keys())

print("Class names:", class_names)

# Evaluate on test data

test_loss, test_acc = model.evaluate(test_ds)

print(f"Test accuracy: {test_acc:.2f}")

# Predict on a single image (replace with your image path)

import os

test_image_path = r'D:\python\python\testing\cat3.jpeg'  # Example cat image

if os.path.exists(test_image_path):

    img = image.load_img(test_image_path, target_size=(224,224))

    img_array = image.img_to_array(img)

    img_array = np.expand_dims(img_array, axis=0) / 255.0  # Normalize like training

    prediction = model.predict(img_array)

    predicted_class = class_names[int(prediction[0] > 0.5)]  # Threshold for binary

    print(f"Predicted class: {predicted_class}")

else:

    print("Test image not found. Update the path to a valid image.")
```
