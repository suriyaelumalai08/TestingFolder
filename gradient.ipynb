```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(0)

n = 200
X = np.linspace(-3, 3, n).reshape(-1, 1)
y = 2 * X.ravel() - 0.3 + 0.7 * np.random.randn(n)

w, b = 0.0, 0.0
lr = 0.05
epochs = 200
losses = []

for _ in range(epochs):
    preds = w * X.ravel() + b
    err = preds - y
    loss = (err**2).mean() / 2
    losses.append(loss)
    dw = (X.ravel() * err).mean()
    db = err.mean()
    w -= lr * dw
    b -= lr * db

plt.figure()
plt.plot(losses, color="green")
plt.title("Linear Regression — Loss (MSE)")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.show()

# plt.figure()
# plt.scatter(X, y, color="lightgreen")
# idx = np.argsort(X.ravel())
# plt.plot(X[idx], (w * X + b)[idx], color="green", linewidth=2)
# plt.title("Linear Regression — Fitted Line")
# plt.xlabel("x")
# plt.ylabel("y")
# plt.grid(True)
# plt.show()
```

```python
import numpy as np

# dataset
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])

y = np.array([[0],[1],[1],[0]])

# weights
W1 = np.random.randn(2,2)
b1 = np.zeros((1,2))

W2 = np.random.randn(2,1)
b2 = np.zeros((1,1))

# functions
def relu(x):
    return np.maximum(0,x)

def relu_derivative(x):
    return x > 0

def sigmoid(x):
    return 1/(1+np.exp(-x))

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s*(1-s)

lr = 0.1

for i in range(10000):

    # forward
    z1 = np.dot(X,W1) + b1
    a1 = relu(z1)

    z2 = np.dot(a1,W2) + b2
    y_pred = sigmoid(z2)

    loss = np.mean((y-y_pred)**2)

    # backprop
    dz2 = (y_pred - y) * sigmoid_derivative(z2)

    dW2 = np.dot(a1.T,dz2)
    db2 = np.sum(dz2,axis=0)

    dz1 = np.dot(dz2,W2.T) * relu_derivative(z1)

    dW1 = np.dot(X.T,dz1)
    db1 = np.sum(dz1,axis=0)

    # update
    W2 -= lr*dW2
    b2 -= lr*db2

    W1 -= lr*dW1
    b1 -= lr*db1

    if i % 1000 == 0:
        print("loss:",loss)
```

```python
from abc import ABC,abstractmethod

class A(ABC):
    @abstractmethod
    def __init__(self):
        pass
    @abstractmethod
    def showA(self):
        pass
    def showA2(self):
        print("class of A2")
    @abstractmethod
    def abstract(self):
        pass
    @abstractmethod
    def func(self):
        pass


class B(A):
    def __init__(self,data):
        self.data=data
        print(self.data)
    def showA(self):
        print('Class B showA using abstarct.')
    def abstract(self):
        return "abstract is working"
    def func(self):
        return "working in func"
    

c=B(10)
c.showA()
c.abstract()
```
